## 8. Matplotlib
### (a)

```python
x = np.arange(0, 10)
y = x ** 2
```

```python
plt.figure()
plt.plot(x, y)
plt.title("y = x^2")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```

### (b)

```python
y1 = x
y2 = x ** 2
y3 = x ** 3
```

```python
plt.figure()
plt.plot(x, y1, linestyle='--', label='y = x')
plt.plot(x, y2, linestyle='-', label='y = x^2')
plt.plot(x, y3, linestyle=':', label='y = x^3')

plt.legend()
plt.title("Multiple Functions")
plt.xlabel("x")
plt.ylabel("y")

plt.text(2, 200, r"$y = x^3$", fontsize=12)
plt.show()
```

### (c)

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(x, y1)
axes[0].set_title("y = x", fontsize=20)
axes[0].set_xlabel("x", fontsize=8)
axes[0].set_ylabel("y", fontsize=16)

axes[1].plot(x, y2)
axes[1].set_title("y = x^2", fontsize=30)
axes[1].set_xlabel("x", fontsize=10)
axes[1].set_ylabel("y", fontsize=12)

plt.tight_layout()
plt.show()
```

### (d)

```python
plt.figure()
plt.plot(x, y2)

ax = plt.gca()
ax.set_title("Using gca()")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.grid(True)

plt.show()
```

## 9. Seaborn
### (b)

```python
sns.set(style="whitegrid")

sns.histplot(df["salary"], bins=30)
plt.title("Distribution of Player Salaries")
plt.show()
```

### (c)

```python
sns.lmplot(
    data=df,
    x="yearID",
    y="salary",
    scatter_kws={"alpha": 0.3}
)
plt.title("Salary vs Year")
plt.show()
```

```python
sns.catplot(
    data=df,
    x="teamID",
    y="salary",
    kind="box",
    height=5,
    aspect=2
)
plt.xticks(rotation=90)
plt.title("Salary Distribution by Team")
plt.show()
```

```python
sns.relplot(
    data=df,
    x="yearID",
    y="salary",
    kind="scatter",
    alpha=0.4
)
plt.show()
```

### (d)

```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x="teamID", y="salary")
plt.xticks(rotation=90)
plt.title("Team Salary Boxplot")
plt.show()
```

### (e)

```python
sns.pairplot(df[["yearID", "salary"]])
plt.show()
```

```python
sns.jointplot(
    data=df,
    x="yearID",
    y="salary",
    kind="scatter"
)
plt.show()
```
